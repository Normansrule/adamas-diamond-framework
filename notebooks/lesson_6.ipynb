```python
n, steps = R.WORKLOADS['Materials simulation (1,000 logical, 10⁹ steps)']
for tr in (100, 1000):
    e = R.estimate(n, steps, tr)
    print(tr, 'µs readout:', e.distance, f'{e.physical_qubits:.2e} qubits', round(e.runtime_hours / 24), 'days')
```
