```python
from adamas import converter
```

```python
for m in ('Si', '4H-SiC', 'GaN', 'Diamond'):
    s = converter.material_switch(m, 1200, 800, 300, 20e3)
    print(f'{m:8s} {s.p_total_w:7.1f} W per switch')
```

```python
s = converter.material_switch('Diamond', 1200, 800, 300, 20e3, measured='saha2022')
print('measured 2022 diamond device, scaled:', round(s.p_total_w, 1), 'W')
```
